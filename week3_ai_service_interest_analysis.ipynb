```python
topic = "나의 AI 서비스 관심도 분석"
print(topic)
```

```python
ai_services = [
    "실시간 번역 AI",
    "이미지 생성 AI",
    "코딩 보조 AI",
    "문서 작성 AI",
    "일정 관리 AI",
    "여행 계획 AI",
    "검색형 AI"
]

print(ai_services)
```

```python
interest_scores = [6, 6, 8, 9, 10, 8, 9]

print(interest_scores)
```

```python
total_score = sum(interest_scores)
average_score = total_score / len(interest_scores)

print("관심도 합계:", total_score)
print("관심도 평균:", average_score)
```

```python
max_score = max(interest_scores)
max_index = interest_scores.index(max_score)

print("가장 높은 관심도:", max_score)
print("가장 관심 있는 AI 서비스:", ai_services[max_index])
```

```python
import matplotlib.pyplot as plt

plt.bar(ai_services, interest_scores)

plt.title("AI 서비스 관심도")
plt.xlabel("AI 서비스")
plt.ylabel("관심도 점수")

plt.xticks(rotation=45)
plt.show()
```
